# Project: Mini Sleep Lab

This project simulates EEG data from scratch to analyze and classify sleep stages based on their dominant frequency bands in 30-second epochs:
- **Awake/relaxed:** Dominant Alpha waves (~10 Hz), small amplitude
- **Light sleep:** Dominant Theta waves (~6 Hz), medium amplitude
- **Deep sleep:** Dominant Delta waves (~2 Hz), large amplitude

## Stage 1: NumPy and Plotting

In this stage, we generate a baseline 30-second clean sine wave (at 2 Hz) sampled at 100 Hz, introduce normal Gaussian noise, and visualize the noisy signal.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Parameters
fs = 100  # Sampling frequency (Hz)
duration = 30  # Duration in seconds
freq = 2  # Signal frequency

# 1. Create time array
t = np.arange(0, duration, 1/fs, dtype=float)

# 2. Build 2 Hz clean sine wave
clean_wave = np.sin(2 * np.pi * freq * t)

# 3. Generate noisy signals at different scales and plot
fig, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True, sharey=True)
for ax, scale in zip(axes, [0.1, 0.5, 2]):
    noise = np.random.normal(loc=0, scale=scale, size=len(t))
    signal = clean_wave + noise
    ax.plot(t[:200], signal[:200], label='Noisy Signal')
    ax.plot(t[:200], clean_wave[:200], 'r--', label='Clean Wave', alpha=0.8)
    ax.set_title(f"Noise Scale = {scale}")
    ax.legend(loc='upper right')
    ax.set_ylabel("Amplitude")

axes[-1].set_xlabel("Time (seconds)")
plt.tight_layout()
plt.show()
```

## Stage 2: Simulating Sleep Stages

Here, we define a functional epoch generator that produces simulated EEG signals based on the designated sleep stage, and construct a synthetic dataset of 60 random epochs.

```python
np.random.seed(42)

def generate_epoch(stage, fs=100, duration=30):
    t = np.linspace(0, duration, fs * duration, endpoint=False)
    # frequency, amplitude parameters per stage
    params = {"awake": (10, 1), "light": (6, 1.5), "deep": (2, 3)}
    frequency, amplitude = params[stage]

    clean_wave = amplitude * np.sin(2 * np.pi * frequency * t)
    noise = np.random.normal(loc=0, scale=0.5, size=len(t))
    return clean_wave + noise

# Generate 60 epochs with random stages
stages = ['awake', 'light', 'deep']
labels = np.random.choice(stages, size=60)
data = np.array([generate_epoch(label) for label in labels])

# Plot an example of each stage
fig, axes = plt.subplots(3, 1, figsize=(10, 6), sharex=True, sharey=True)
for ax, st in zip(axes, stages):
    signal = generate_epoch(st)
    ax.plot(t[:200], signal[:200])
    ax.set_title(f"Simulated Stage: {st.capitalize()}")
    ax.set_ylabel("Amplitude")

axes[-1].set_xlabel("Time (seconds)")
plt.tight_layout()
plt.show()
```

## Stage 3: Feature Extraction and Pandas DataFrame

We define a Fast Fourier Transform (FFT) helper function to compute power within specific frequency bands (Delta, Theta, Alpha) and construct a summary DataFrame.

```python
def band_power(epoch, fs, low, high):
    power = np.abs(np.fft.rfft(epoch)) ** 2
    freqs = np.fft.rfftfreq(len(epoch), d=1/fs)
    mask = (freqs >= low) & (freqs < high)
    return power[mask].sum()

rows = []
for i in range(len(data)):
    epoch = data[i]
    rows.append({
        'epoch': i,
        'stage': labels[i],
        'std': np.std(epoch),
        'delta': band_power(epoch, fs, 0.5, 4),
        'theta': band_power(epoch, fs, 4, 8),
        'alpha': band_power(epoch, fs, 8, 13)
    })

df = pd.DataFrame(rows).set_index('epoch')
display(df.head(10))
```

## Stage 4: Analysis and Visualizations

We perform group analyses and construct visualizations to verify that our frequency-based features can successfully separate the sleep stages.

```python
# 1. Groupby Means
print("Mean Band Power per Stage:")
display(df.groupby("stage")[["delta", "theta", "alpha"]].mean())

# 2. Bar Plot of Aggregated Band Powers per Stage
fig, ax = plt.subplots(figsize=(8, 5))
df.groupby("stage")[["delta", "theta", "alpha"]].mean().plot(kind="bar", ax=ax)
ax.set_title("Mean Band Power per Sleep Stage")
ax.set_ylabel("Power (Log Scale)")
ax.set_yscale("log")
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()

# 3. Boxplot for Delta Power
plt.figure(figsize=(7, 5))
df.boxplot(column="delta", by="stage")
plt.yscale("log")
plt.title("Delta Power Distribution by Sleep Stage")
plt.suptitle("") # Remove default pandas title
plt.ylabel("Delta Power (Log Scale)")
plt.show()

# 4. Scatter Plot: Delta vs Alpha Power
fig, ax = plt.subplots(figsize=(7, 5))
for s in stages:
    sub = df[df["stage"] == s]
    ax.scatter(sub["delta"], sub["alpha"], label=s, edgecolors='k', alpha=0.8)
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Delta Power (0.5 - 4 Hz)")
ax.set_ylabel("Alpha Power (8 - 13 Hz)")
ax.set_title("Sleep Stage Clustering (Delta vs Alpha)")
ax.legend()
plt.show()
```

## Stage 5: Rule-Based Stage Prediction

Using the extracted features, we predict the sleep stages by choosing the band with the maximum power and map them back to evaluate our accuracy.

```python
# Predict stage based on the dominant frequency band
pred = df[["delta", "theta", "alpha"]].idxmax(axis=1)
map_pred = pred.map({"delta": "deep", "theta": "light", "alpha": "awake"})

accuracy = (map_pred == df["stage"]).mean()
print(f"Classification Accuracy: {accuracy * 100:.2f}%")
```
